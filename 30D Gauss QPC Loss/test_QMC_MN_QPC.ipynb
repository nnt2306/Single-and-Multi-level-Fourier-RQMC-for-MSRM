```python
# =========================
# Print helpers
# =========================
def print_1d_history(vals, title, show=6,ind = "damping"):
    print(title)
    print("-" * len(title))
    n = len(vals)
    head = min(show, n)
    for it, d in enumerate(vals[:head]):
        print(f"iter {it:3d} | {ind} = {d:10.4g}")
    if n > show:
        print(f"... ({n - show} more iterations not shown)")

def print_2d_history(vals, title, show=6):
    print(title)
    print("-" * len(title))
    n = len(vals)
    head = min(show, n)
    for it, tup in enumerate(vals[:head]):
        tup_str = "(" + ", ".join(f"{x:10.4g}" for x in tup) + ")"
        print(f"iter {it:3d} | tuple = {tup_str}")
    if n > show:
        print(f"... ({n - show} more iterations not shown)")
```

## Single-level Fourier-RQMC

```python
# =========================
# Build single-level loss object, here for numerical convenience we fix the damping at the initial level
# =========================
loss_qmc_Fou_qpc_30D = RQMC_Fou_MN_qpc(
    v_mu=mu_30,
    v_sigma=Sigma_30,
    N_sobol=N_sobol,
    m_shift=m_shift,
    alpha=alpha,
    c=c,
    epsilon=epsilon_sing,
    fix_damping=True,
)


# =========================
# Internal bookkeeping helpers at each iteration
# =========================
def _set_current_m(m):
    """Store current iterate information"""
    m = np.asarray(m, dtype=float)
    loss_qmc_Fou_qpc_30D._current_m = m

    pending_cov = loss_qmc_Fou_qpc_30D._pending_jac_cov
               
    if pending_cov is not None:
        loss_qmc_Fou_qpc_30D._jac_cov_history.append(
            pending_cov.copy()
        )
        loss_qmc_Fou_qpc_30D._pending_jac_cov = None

## Objective: \sum_{k=1}^{d} m_k
def objective_wrapper(m):
    m = np.asarray(m, dtype=float)
    loss_qmc_Fou_qpc_30D._current_m = m.copy()
    return loss_qmc_Fou_qpc_30D.objective(m)


## Gradient of the objective
def jac_wrapper(m):
    m = np.asarray(m, dtype=float)
    loss_qmc_Fou_qpc_30D._current_m = m.copy()
    return loss_qmc_Fou_qpc_30D.objective_jac(m)


def scipy_callback(m):
    _set_current_m(m)


# =========================
# Warm-up evaluation at initial point of the optimization
# =========================
loss_qmc_Fou_qpc_30D._current_m = np.asarray(guess, dtype=float)
loss_qmc_Fou_qpc_30D.shortfall_risk(guess)
loss_qmc_Fou_qpc_30D.shortfall_risk_jac(guess)
_set_current_m(guess)

# =========================
# Constraint E[l(X - m)] and its gradient E[∇l(X - m)]
# =========================
cons_qmc_exact = {
    "type": "ineq",
    "fun": lambda x: loss_qmc_Fou_qpc_30D.ineq_constraint(x),
    "jac": lambda x: loss_qmc_Fou_qpc_30D.ineq_constraint_jac(x),
}

# =========================
# Optimization
# =========================
res = minimize(
    fun=objective_wrapper,
    x0=guess,
    jac=jac_wrapper,
    constraints=cons_qmc_exact,
    method="SLSQP",
    options={"maxiter": maxiter, "ftol": 1e-4},
    callback=scipy_callback,
)

print(f"Number of iterations: {res.nit}")

# =========================
# Final solution + relative error
# =========================
max_qmc_err = loss_qmc_Fou_qpc_30D.statistical_error_sol_RQMC(res.x)
rel_err = max_qmc_err / np.max(res.x)

print("\nFinal result")
print("------------")
print(f"Single Fourier-RQMC solution: {res.x}")
print(f"Relative statistical error : {rel_err:.6g}")
```

## Multilevel Fourier-RQMC

```python
# =========================
# Build multilevel Fourier-RQMC loss object
# =========================
loss_qmc_CV_Fou_qpc_30D = RQMC_CV_Fou_MN_qpc(
    v_mu=mu_30,
    v_sigma=Sigma_30,
    N_sobol=N_sobol,
    m_shift=m_shift,
    alpha=alpha,
    c=c,
    epsilon= epsilon_mult,
)

# =========================
# Internal bookkeeping helpers at each iteration
# =========================
#
N_sobol_iter = [] # List to keep track for number of Sobol points
count = 0 # This will use to keep track when we enter the local convergence region
def _set_current_m(m):
    """Store current iterate."""
    global count
    m = np.asarray(m, dtype=float)
    loss_qmc_CV_Fou_qpc_30D._current_m = m
    
    # The commit option in this case help us to store only succesful optimization iteration
    loss_qmc_CV_Fou_qpc_30D.shortfall_risk(m, commit=True)
    loss_qmc_CV_Fou_qpc_30D.shortfall_risk_jac(m, commit=True)
    pending_cov =  loss_qmc_CV_Fou_qpc_30D._pending_jac_cov
                
    if pending_cov is not None:
           loss_qmc_CV_Fou_qpc_30D._jac_cov_history.append(
            pending_cov.copy()
        )
           loss_qmc_CV_Fou_qpc_30D._pending_jac_cov = None
        
    count += 1

    ## Adaptive choice of Sobol point when entering local convergence region
    N_sobol_iter.append(loss_qmc_CV_Fou_qpc_30D._N_current_loss)
    if count > 4:
        loss_qmc_CV_Fou_qpc_30D._divide_sobol(mult_factor=1/2)
        loss_qmc_CV_Fou_qpc_30D._divide_sobol(mult_factor=1/2,grad= True)
    
    loss_qmc_CV_Fou_qpc_30D.record_qmc_stats()

## Objective: \sum_{k=1}^{d} m_k
def objective_wrapper(m):
    m = np.asarray(m, dtype=float)
    loss_qmc_CV_Fou_qpc_30D._current_m = m.copy()
    return loss_qmc_CV_Fou_qpc_30D.objective(m)

## Gradient of the objective 
def jac_wrapper(m):
    m = np.asarray(m, dtype=float)
    loss_qmc_CV_Fou_qpc_30D._current_m = m.copy()
    return loss_qmc_CV_Fou_qpc_30D.objective_jac(m)

def scipy_callback(m):
    _set_current_m(m)

# =========================
# Warm-up evaluation at initial point of the optimization
# =========================
loss_qmc_CV_Fou_qpc_30D._current_m = np.asarray(guess, dtype=float)
loss_qmc_CV_Fou_qpc_30D.shortfall_risk(guess)
loss_qmc_CV_Fou_qpc_30D.shortfall_risk_jac(guess)
_set_current_m(guess)

# =========================
# Constraint E[l(X - m)] and its gradient E[∇l(X - m)]
# =========================
cons_qmc_exact = {
    "type": "ineq",
    "fun": lambda x: loss_qmc_CV_Fou_qpc_30D.ineq_constraint(x),
    "jac": lambda x: loss_qmc_CV_Fou_qpc_30D.ineq_constraint_jac(x),
}

# =========================
# Optimization
# =========================
res = minimize(
    fun=objective_wrapper,
    x0=guess,
    jac=jac_wrapper,
    constraints=cons_qmc_exact,
    method="SLSQP",
    options={"maxiter": maxiter, "ftol": 1e-4},
    callback=scipy_callback,
)
print(f"Number of iterations: {res.nit}")
print_1d_history(N_sobol_iter,"Number of Sobol points",show = 10,ind = "N_Sobol")


# Final solution + relative error
# =========================
max_qmc_err = loss_qmc_CV_Fou_qpc_30D.statistical_error_sol_RQMC(res.x,loc_index=6)
rel_err = max_qmc_err / np.max(res.x)

print("\nFinal result")
print("------------")
print(f"Multilevel Fourier-RQMC solution: {res.x}")
print(f"Relative statistical error : {rel_err:.6g}")

```
